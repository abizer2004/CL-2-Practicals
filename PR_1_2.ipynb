implementation of stemming words using NLTK from an array of words.

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
```

```python
ps = PorterStemmer()
# choose some words to be stemmed
words = ["program", "programs", "programmer", "programming", "programmers"]
```

```python
for w in words:
    print(w, " : ", ps.stem(w))
```

implementation of stemming words using NLTK from sentences.

```python
ps = PorterStemmer()
sentence = "Programmers program with programming languages"
words = word_tokenize(sentence)
```

```python
for w in words:
    print(w, " : ", ps.stem(w))
```
